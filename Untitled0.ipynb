```python
from pathlib import Path
import re

notebook_path = Path(
    "/content/drive/MyDrive/study/vision-model-playground/Untitled0.ipynb"
)

notebook_text = notebook_path.read_text(encoding="utf-8")

token_patterns = [
    r"github_pat_[A-Za-z0-9_]{20,}",
    r"gh[pousr]_[A-Za-z0-9]{20,}",
]

detected_count = sum(
    len(re.findall(pattern, notebook_text))
    for pattern in token_patterns
)

print("검출된 GitHub 토큰 패턴 수:", detected_count)
```
